## 0. How to Complete These Exercises

This notebook runs on the **evcxr** kernel, the same one the example notebook
uses. Three consequences shape how you work:

1. **No `fn main`.** The body of a cell *is* the body of `main`. Type statements
   directly.
2. **State persists.** A value bound in one cell is still visible in every cell
   below it, so the sections build on each other.
3. **Items are order-independent.** `fn` definitions, `const`s and `use` lines
   may be used above the cell that defines them — handy, but it means a
   notebook that runs top-to-bottom is not proof that it would compile as a
   normal crate.

A fourth consequence is worth stating explicitly, because it bites:

> **Item names must be unique across the whole notebook.** Every cell shares one
> namespace, so two functions with the same name in different sections are a
> redefinition error. If the compiler tells you a name "is defined multiple
> times", it is pointing at an *earlier* cell. Rename the new one.

### The stub convention

Each exercise cell declares a signature, a doc comment explaining the contract,
and a block of `assert_eq!`/`assert!` calls. The body you must fill in is
`todo!()`, which compiles but **panics** when executed:

```text
thread '<main>' panicked at ... not yet implemented
```

Replace the `todo!()` (or the `# TODO` comment) with real code and re-run the
cell. An `Exercise N.M passed` line means you are done.

### Expect warnings until you finish

Every unsolved stub whose parameters you have not yet touched produces an
`unused variable` warning, so a fresh notebook shows a long list of them. That
is normal and self-cleaning: each warning disappears when you implement the
body. If a *new* diagnostic appears that is not in that list, it is yours.

Two practical notes:

- Re-running a cell that *defines* a function redefines it. On the rare
  `cannot define ... twice` error, restart the kernel and run the cells again.
- Errors reported by the kernel are plain `rustc` errors. Read them from the
  top, fix the first one, and re-run — the later ones are usually knock-on.

### Resetting

Use *Restart Kernel* when state gets confusing. Then run the cells you need,
top to bottom.

**See**: §0 and §15 of the example notebook.

## 2. Bindings and Mutability

### Exercise 2.1 — Opting in to `mut`

**Goal**: change a local binding.

**Info**: Bindings are immutable unless declared `mut`. Forgetting it yields

```text
error[E0384]: cannot assign twice to immutable variable
```

**Objectives**

- return `n` with 1 added, three times over
- declare the binding `mut` — nowhere else will do

**See**: §2.1 of the example notebook.

**Hint**: `n += 1` is shorthand for `n = n + 1`. Three statements in a row is
fine, or a `for` loop over `0..3`.


````rust
/// Add 1 to `n` three times, then return it.
///
/// TODO: bind `n` as `mut`, then increment it three times.
use std::ops::Range;
fn thrice_incremented(mut n: i32) -> i32 {
    let range = Range{start: 0, end: 3};
    // let range = 0..3;
    for i in range {
        n = n + 1;
    }
    return n;
}

// --- check (do not modify) ---
assert_eq!(thrice_incremented(0), 3);
assert_eq!(thrice_incremented(-3), 0);
assert_eq!(thrice_incremented(10), 13);
println!("Exercise 2.1 passed");
````

## 3. Shadowing

### Exercise 3.1 — Rebinding instead of mutating

**Goal**: transform a value by shadowing it.

**Info**: `let n = n * 2;` creates a *new* binding that hides the old one. The
original is untouched, and no `mut` is needed. Shadowing works in the same scope
and inside inner blocks.

**Objectives**

- shadow `n` so that it holds double its incoming value
- return `"big"` when the doubled value is **greater than** 10, else `"small"`
- use no `mut` anywhere

**See**: §3.1 of the example notebook.

**Hint**: the `if` can be a tail expression:
`if n > 10 { "big" } else { "small" }`.


````rust
/// Double `n` by shadowing it, then report whether the result exceeds 10.
///
/// TODO: `let n = n * 2;` first, then answer with an `if` expression.
fn doubled_is_big(n: i32) -> &'static str {
    // if (n > 10) { return "big"; }
    // else { return "small"; }
    return (if (n * 2 > 10) {"big"} else {"small"});
}

// --- check (do not modify) ---
assert_eq!(doubled_is_big(6), "big");
assert_eq!(doubled_is_big(5), "small");
assert_eq!(doubled_is_big(-1), "small");
println!("Exercise 3.1 passed");
````

### Exercise 3.2 — Shadowing across a type change

**Goal**: shadow a `&str` into an `i32`.

**Info**: Shadowing may change the *type* as well as the value, which is what
makes `let x = "42"; let x: i32 = x.parse().unwrap();` legal. This is the
idiomatic way to convert text into a number without inventing names like
`x_str` and `x_num`.

**Objectives**

- parse `text` into an `i32` and add 100
- shadow the binding rather than introducing a new name

**See**: §3.1 and §5.2 of the example notebook.

**Hint**: `text.parse()` needs to know the target type — the annotation
`let n: i32 = ...` supplies it. `.unwrap()` is acceptable here because the
checks only feed valid input.


````rust
/// Parse `text` as an integer, then return it with 100 added.
///
/// TODO: shadow `text` into an `i32`, then return the sum.
fn offset(text: &str) -> i32 {
    let mut text: i32 = text.parse().unwrap();
    return text + 100;
}

// --- check (do not modify) ---
assert_eq!(offset("42"), 142);
assert_eq!(offset("-2"), 98);
assert_eq!(offset("0"), 100);
println!("Exercise 3.2 passed");
````

### Exercise 4.2 — `char` is not a byte

**Goal**: distinguish characters from bytes.

**Info**: A `char` is a Unicode **scalar value**, always 4 bytes in memory, but
between 1 and 4 bytes when encoded as UTF-8. `char::len_utf8()` gives the
encoded width; `std::mem::size_of::<char>()` gives the in-memory width.

**Objectives**

- `is_upper_letter` returns `true` only for `'A'`–`'Z'`
- `utf8_width` returns 1 for `'a'`, 2 for `'ß'`, 3 for `'€'`

**See**: §4.1 and §9.1 of the example notebook, and
[Rust std docs: `char`](https://doc.rust-lang.org/std/primitive.char.html).

**Hint**: `char` has `is_ascii_uppercase()`, `is_alphabetic()`,
`is_numeric()` — all return `bool`. In UTF-8 everything outside ASCII needs at
least 2 bytes.


````rust
/// Return `true` when `c` is an uppercase ASCII letter.
///
/// TODO: `char` methods will do this in one call.
fn is_upper_letter(c: char) -> bool {
    return c.is_ascii_uppercase();
}

/// Return the number of bytes `c` occupies in UTF-8.
///
/// TODO: one `char` method gives this directly.
fn utf8_width(c: char) -> usize {
    return c.len_utf8();
}

// --- check (do not modify) ---
assert!(is_upper_letter('A'));
assert!(!is_upper_letter('a'));
assert!(!is_upper_letter('5'));
assert!(!is_upper_letter('Ä'));
assert_eq!(utf8_width('a'), 1);
assert_eq!(utf8_width('ß'), 2);
assert_eq!(utf8_width('€'), 3);
assert_eq!(utf8_width('😀'), 4);
println!("Exercise 4.2 passed");
````

## 5. Type Inference and Annotations

### Exercise 5.1 — Make the literal fit

**Goal**: supply the type that a bare literal does not carry.

**Info**: An unsuffixed integer literal defaults to `i32`; an unsuffixed decimal
literal defaults to `f64`. `7` therefore cannot be returned from a `-> u16`
function without help.

**Objectives**

- return 7 as a `u16`
- return 0.5 as an `f32`

**See**: §5.1 and §6.1 of the example notebook.

**Hint**: either suffix the literal (`7u16`, `0.5f32`) or annotate the
expression with the target type. Both are equally idiomatic.


````rust
/// Return 7 as a `u16`.
///
/// TODO: the literal alone would be an `i32`.
fn seven_as_u16() -> u16 {
    // return 7_u16;
    let n: u16 = 7;
    return n;
}

/// Return a value that is exactly one half, as an `f32`.
///
/// TODO: the literal alone would be an `f64`.
fn half_f32() -> f32 {
    // return 0.5_f32;
    let n: f32 = 0.5;
    return n;
}

// --- check (do not modify) ---
assert_eq!(seven_as_u16(), 7u16);
assert_eq!(half_f32(), 0.5f32);
println!("Exercise 5.1 passed");
````

### Exercise 5.2 — Where inference needs help

**Goal**: make a generic call tell the compiler what you want.

**Info**: `str::parse` is generic over the *target* type. A function's declared
return type is often enough of a hint — the body's value must have that type, so
inference flows backwards. When it is not enough, use the turbofish:
`text.parse::<i32>()`.

**Objectives**

- `parse_i64` relies on the return type to disambiguate
- `parse_generic` is generic, so *the caller* chooses, via `parse::<T>()`

**See**: §5.2 of the example notebook, and
[Rust std docs: `FromStr`](https://doc.rust-lang.org/std/str/trait.FromStr.html).

**Hint**: for the generic version you need two bounds — `T: FromStr` on the
type, and `T::Err: Debug` because `.unwrap()` must be able to print the failure.


````rust
/// Parse `text` and return it as an `i64`.
///
/// TODO: the declared return type is the hint `parse` needs.
fn parse_i64(text: &str) -> i64 {
    return text.parse::<i64>().unwrap();
}

/// Parse `text` into whatever the caller asks for.
///
/// TODO: `text.parse::<T>().unwrap()` is the shape; the bounds are given.
fn parse_generic<T>(text: &str) -> T
where
    T: std::str::FromStr,
    T::Err: std::fmt::Debug,
{
    return text.parse::<T>().unwrap();
}

// --- check (do not modify) ---
assert_eq!(parse_i64("1234"), 1234i64);
assert_eq!(parse_i64("-8"), -8i64);
assert_eq!(parse_generic::<i32>("-9"), -9i32);
assert_eq!(parse_generic::<u8>("200"), 200u8);
assert_eq!(parse_generic::<f64>("1.5"), 1.5f64);
println!("Exercise 5.2 passed");
````

## 7. Operators, Casting and Safe Arithmetic

### Exercise 7.1 — What `as` really does

**Goal**: internalise that `as` truncates bits without warning.

**Objectives**

- return `value` reduced to a `u8` the way `as` does it

**See**: §7.2 of the example notebook.

**Hint**: one `as` is the whole body.

````rust
/// Return `value` reduced to a `u8` the way `as` does it.
///
/// TODO: a single `as` cast.
fn as_u8(value: i32) -> u8 {
    return value as u8;
}

// --- check (do not modify) ---
assert_eq!(as_u8(255), 255);
assert_eq!(as_u8(256), 0);
assert_eq!(as_u8(300), 44);
assert_eq!(as_u8(-1), 255);
println!("Exercise 7.1 passed");
````

### Exercise 7.2 — Float to integer

**Goal**: choose between truncation and rounding deliberately.

**Info**: `as` between a float and an integer **truncates toward zero**. To
round, the float must first be turned into a whole number with `.round()`,
`.floor()` or `.ceil()`. Note that `(-2.5).round()` is `-3`: ties round *away
from zero*, not to even. Out-of-range values saturate rather than wrap, so a cast
can never panic — it can only surprise you.

**Objectives**

- `truncated` throws away the fraction
- `rounded` goes to the nearest integer

**See**: §7.2 and §8.3 of the example notebook.

**Hint**: `.round()` returns an `f64`, so the cast is still needed:
`x.round() as i32`.

````rust
/// Convert `x` to an `i32` by truncating toward zero.
///
/// TODO: `as` already truncates.
fn truncated(x: f64) -> i32 {
    return x as i32;
}

/// Convert `x` to an `i32` by rounding to the nearest integer.
///
/// TODO: round first, then cast.
fn rounded(x: f64) -> i32 {
    return x.round() as i32;
}

// --- check (do not modify) ---
assert_eq!(truncated(2.9), 2);
assert_eq!(truncated(-2.9), -2);
assert_eq!(rounded(2.5), 3);
assert_eq!(rounded(-2.5), -3);
println!("Exercise 7.2 passed");
````

### Exercise 7.3 — The three overflow policies

**Goal**: name the arithmetic families you will reach for in real code.

**Info**: `u8` arithmetic **panics in debug builds** and wraps in release builds,
which is the worst of both worlds: the bug hides exactly where you are not
looking. Pick a policy explicitly instead.

| Family | On overflow |
|---|---|
| `checked_*` | returns `None` |
| `wrapping_*` | wraps modulo 2^n |
| `saturating_*` | clamps at `MIN` or `MAX` |
| `overflowing_*` | returns `(value, did_overflow)` |

**Objectives**

- `wrap_add(250, 10)` is `4`
- `sat_add(250, 10)` is `255`
- `checked_add(250, 10)` is `None`

**See**: §7.3 of the example notebook, and
[Rust std docs: `u8`](https://doc.rust-lang.org/std/primitive.u8.html).

**Hint**: all three are methods on `u8` itself, and all three have the same
`(self, rhs)` shape.

````rust
/// Add `delta` to `start`, wrapping modulo 256.
///
/// TODO: one method call.
fn wrap_add(start: u8, delta: u8) -> u8 {
    return start.wrapping_add(delta);
}

/// Add `delta` to `start`, clamping at `u8::MAX`.
///
/// TODO: one method call.
fn sat_add(start: u8, delta: u8) -> u8 {
    return start.saturating_add(delta);
}

/// Add `delta` to `start`, returning `None` on overflow.
///
/// TODO: one method call.
fn checked_add(start: u8, delta: u8) -> Option<u8> {
    return start.checked_add(delta);
}

// --- check (do not modify) ---
assert_eq!(wrap_add(250, 10), 4);
assert_eq!(sat_add(250, 10), 255);
assert_eq!(checked_add(250, 10), None);
assert_eq!(checked_add(1, 2), Some(3));
assert_eq!(wrap_add(1, 2), 3);
println!("Exercise 7.3 passed");
````

### Exercise 7.4 — Signs, division and bits

**Goal**: read Rust's integer division and bitwise operators without guessing.

**Info**: `/` truncates **toward zero** and `%` keeps the sign of the **left**
operand, so `-7 / 3` is `-2` and `-7 % 3` is `-1`. Neither one floors. When you
want the mathematical remainder — the one that is always non-negative for a
positive divisor — say so explicitly with `div_euclid` and `rem_euclid`. The
sign convention of `%` is the single most common source of off-by-one bugs in
index arithmetic, so it is worth *knowing* rather than *checking*.

Bitwise operators act on every bit at once and do not short-circuit. Note too
that in Rust `&` binds **tighter** than `==` — the opposite of C, where `a & b
== c` means `a & (b == c)` and is a classic bug.

**Objectives**

- `trunc_rem(-7, 3)` is `-1`, and `trunc_rem(7, -3)` is `1`
- `euclid_rem(-7, 3)` is `2` — non-negative, unlike `%`
- `low_nibble(0xAB)` is `0x0B`
- `high_nibble(0xAB)` is `0x0A`

**See**: §7 of the example notebook, and
[Rust std docs: `i32::rem_euclid`](https://doc.rust-lang.org/std/primitive.i32.html#method.rem_euclid).

**Hint**: the first two are the `%` operator. The third is `.rem_euclid()`. For
the nibbles, `0x0F` keeps the bottom four bits and `>> 4` slides the top four
down; no parentheses are needed, since shifts and masks bind tighter than
anything you have around them.

## 8. Floating-Point Notes

### Exercise 8.1 — Comparing floats is a policy decision

**Goal**: write the tolerance comparison you will reuse forever.

**Info**: `0.1 + 0.2 != 0.3` because neither decimal has an exact binary
representation. The fix is not a different operator but a **tolerance**: two
floats count as equal when their absolute difference is small. Use `<=`, not
`<`, so that identical inputs always count as equal even with `tolerance == 0`.

**Objectives**

- `approx_eq` compares `|a - b|` against `tolerance`
- never use `==` on the floats

**See**: §8.1 of the example notebook.

**Hint**: `(a - b).abs() <= tolerance`. If you also need to handle huge

here.


````rust
/// Return `true` when `a` and `b` differ by no more than `tolerance`.
///
/// TODO: compare the absolute difference; do not use `==`.
fn approx_eq(a: f64, b: f64, tolerance: f64) -> bool {
    return (a - b).abs() <= tolerance
}

// --- check (do not modify) ---
assert!(approx_eq(0.1 + 0.2, 0.3, 1e-12));
assert!(!approx_eq(0.1 + 0.2, 0.3, 0.0));
assert!(approx_eq(1.0, 1.0, 0.0));
assert!(approx_eq(-2.0, -2.000_000_1, 1e-5));
println!("Exercise 8.1 passed");
````

### Exercise 8.2 — NaN, and the missing `Ord`

**Goal**: work around the one float value that breaks every comparison.

**Info**: `f64` implements `PartialOrd` but **not** `Ord`, because `NaN` is
unordered with respect to everything, including itself. That single fact
removes `.sort()` and `HashMap` keys from the toolbox for floats. The escape
hatch is `f64::total_cmp`, which imposes a total order and pushes `NaN` to the
end.

**Objectives**

- `not_a_number` detects NaN
- `sort_ascending` sorts in place with NaN last

**See**: §8.2 of the example notebook, and
[Rust std docs: `f64::total_cmp`](https://doc.rust-lang.org/std/primitive.f64.html#method.total_cmp).

**Hint**: `x.is_nan()` for the first. For the second, `values` is a
`&mut [f64]`, so `values.sort_by(...)` with the comparison function as the
argument — note that `f64::total_cmp` takes two `&f64`, which is exactly what

````rust
/// Return `true` when `x` is not a number.
///
/// TODO: one method call.
fn not_a_number(x: f64) -> bool {
    return x.is_nan();
}

/// Sort `values` ascending, placing NaN at the end.
///
/// TODO: `f64` has no `Ord`, so `.sort()` will not compile.
fn sort_ascending(values: &mut [f64]) {
    return values.sort_by(f64::total_cmp);
}

// --- check (do not modify) ---
assert!(not_a_number(f64::NAN));
assert!(!not_a_number(1.0));
let mut xs = [3.0, f64::NAN, -1.0, 2.0];
sort_ascending(&mut xs);
// Note: `assert_eq!(xs, [...])` cannot be used here -- NaN != NaN.
assert_eq!(xs[0], -1.0);
assert_eq!(xs[1], 2.0);
assert_eq!(xs[2], 3.0);
assert!(xs[3].is_nan());
println!("Exercise 8.2 passed");
````

### What Comes Next

You now have the vocabulary of a Rust file: bindings, the scalar types, the
format language, tuples and arrays, and the expression/statement distinction
that governs how every function body is read.

Chapter 02 takes the two control-flow forms you have already used informally —
`if` and `for` — and adds the ones you have not: `loop`, `while`, loop labels,
ranges as first-class values, and `match`.

Two habits worth keeping from here on:

- When the compiler disagrees with you, read the **first** error and the
  `= note:` lines under it. The note usually names the fix.
- Prefer an explicit overflow policy (`checked_`, `wrapping_`, `saturating_`) to
  the default panic-in-debug-warn-in-release behaviour.
