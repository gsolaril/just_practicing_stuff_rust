## 0. How to Complete These Exercises

This notebook runs on the same **evcxr** kernel as the example notebook. Three consequences shape how you work:

1. **No `fn main`.** The body of a cell *is* the body of `main`. Type statements directly.
2. **State persists.** A binding from one cell is visible in every cell below it.
3. **Items are order-independent.** A `fn`, a `struct`, an `impl` or a `use` line may be used above the cell that defines it.

A fourth consequence bites harder here than anywhere else in the track, because nearly every exercise defines functions:

> **Item names must be unique across the whole notebook.** Every cell shares one namespace, so two functions with the same name are a redefinition error, even in different sections. If the compiler says a name "is defined multiple times", it is pointing at an *earlier* cell. Rename the new one.

### The stub convention

Each exercise cell declares a signature, a doc comment stating the contract, and a `// --- check (do not modify) ---` block of `assert_eq!`/`assert!` calls. The body you must fill in is `todo!()`, which compiles but **panics** when executed:

```text
thread '<main>' panicked at ... not yet implemented
```

Replace the `todo!()` and re-run the cell. An `Exercise N.M passed` line means you are done. For the two exercises whose stubs are `impl` blocks (§9.1, §10.2), every method body is a separate `todo!()`; the cell passes only when all of them are implemented.

### Borrow errors are expected here

This is the ownership chapter: some checks are written to fail to *compile* if you solve them by borrowing wrongly, and the stubs deliberately invite a first attempt the compiler rejects. When that happens, read the diagnostic (§13 of the example notebook documents all five codes), and change the *approach* — not the signature.

### Expect warnings until you finish

Every unsolved stub whose parameters are untouched emits `unused variable`, so a fresh notebook shows a long list of them. That is normal and self-cleaning: each warning disappears when you implement the body.

Two warning shapes are specific to this notebook, and both are the same self-cleaning mechanism:

- `field ... is never read` and `method ... is never used` appear on the `impl`-stub exercises (§9.1, §9.2, §10.2, §10.4): a stubbed method never reads its struct's fields, so the compiler cannot yet see the field as used. Implementing the method makes both warnings vanish together.
- If a diagnostic appears that is in neither list, it is yours.

**See**: §0 and §13 of the example notebook.
